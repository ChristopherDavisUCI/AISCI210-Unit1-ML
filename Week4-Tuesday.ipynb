* Import the "mpg" dataset from Seaborn.  (You can see other available datasets using the `get_dataset_names` function.)
* Here we will try to predict whether a car has at least 25 mpg using "weight" and "horsepower".
* Add a Boolean column to the DataFrame indicating whether or not the car has at least 25 mpg.  Call this column "target".

```python
import seaborn as sns
```

```python
sns.get_dataset_names()
```

```python
df = sns.load_dataset("mpg")
```

```python
df.info()
```

```python
df = df.dropna(axis=0).copy()
```

* Add a Boolean column to df indicating True if the car has 25mpg or higher.  Call this column "target".

```python
df["target"] = (df["mpg"] >= 25)
```

```python
df
```

* For some tasks, a 10% success rate is good, sometimes it's not.  What would the success rate of a naive baseline prediction be for this dataset?

```python
df["target"].mean()
```

```python
1 - df["target"].mean()
```

```python
(df["target"] == False).mean()
```

Upshot: Can get over 57.5% accuracy just by always predicting False.  In some cases, can get a 98% accuracy doing something similar.  Always should keep this baseline performance in mind when evaluating your Machine Learning model.

* Fit a `LogisticRegression` object using "weight" and "horsepower" for the predictors and using "target" for our target.
* Notice how similar a `LogisticRegression` object from scikit-learn is to a `LinearRegression` object that we saw last week.
* We're going to need to remove some missing values.  Would you expect to use `axis=0` or `axis=1`?

```python
from sklearn.linear_model import LogisticRegression
```

```python
clf = LogisticRegression()
```

```python
df["weight", "horsepower"]
```

```python
df[["weight", "horsepower"]]
```

```python
clf.fit(df[["weight", "horsepower"]], df["target"])
```

* What accuracy do we have here?

```python
(clf.predict(df[["weight", "horsepower"]]) == df["target"]).mean()
```

```python
clf2 = LogisticRegression()
```

```python
clf2.fit(df[["weight", "horsepower"]], df["origin"])
```

```python
clf2.predict(df[["weight", "horsepower"]])
```

```python
clf2.predict(df[["weight", "horsepower"]]) == df["origin"]
```

```python
(clf2.predict(df[["weight", "horsepower"]]) == df["origin"]).mean()
```

```python
clf.score(df[["weight", "horsepower"]], df["target"])
```

```python
clf2.score(df[["weight", "horsepower"]], df["origin"])
```
